```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Exploratory Data Analysis (EDA) - Bar Chart 10 Most Flight Duration (Round Up)
df = pd.read_csv("airlines_flights_data.csv")

# Group by Duration (Ceil)
duration_summary = (
    df.assign(duration_hour=np.ceil(df['duration']))   
      .groupby('duration_hour')
      .size()
      .reset_index(name='total_flight')
      .sort_values(by='total_flight', ascending=False)
      .head(10)
)

# Bar Chart
plt.figure(figsize=(10, 6))
ax = duration_summary.plot(
    x='duration_hour',
    y='total_flight',
    kind='bar',
    legend=False
)
# Add label
for i, value in enumerate(duration_summary['total_flight']):
    plt.text(i, value + 1, str(value), ha='center', va='bottom')
plt.title("10 Most Flight Duration (Ceil)")
plt.xlabel("Duration Hour")
plt.ylabel("Number of Flights")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
